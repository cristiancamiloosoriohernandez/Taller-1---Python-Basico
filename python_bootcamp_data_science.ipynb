2.Objetivo: Usar bucles y condiciones para clasificar edades y buscar nombres en una lista de estudiantes.

*Usa un bucle for para iterar sobre cada edad:

*Si la edad es 25, imprime "¡Encontramos al estudiante de 25 años! Deteniendo el análisis." y termina el bucle con break.

*Si la edad es menor a 10, omite esa iteración con continue.

*Si la edad es 18 o mayor, imprime "Adulto: [edad] años".

*En cualquier otro caso, imprime "Menor: [edad] años".

```python
edades = [12, 17, 8, 15, 22, 9, 30, 25]
```

```python
for e in edades:
  if e == 25:
    print("¡Encontramos al estudiante de 25 años! Deteniendo el análisis.")

  elif e < 10:
    continue

  elif e >= 18:
    print("Adulto:", e, "años")

  else:
    print("Menor:", e, "años")
```

# Actividad 5 - Funciones y Manejo de Errores

1.Crea una función llamada aplicar_descuento_iva que reciba dos parámetros: el precio_producto y el porcentaje_descuento. Tu función debe realizar los siguientes procesos internos:

*Restar el porcentaje de descuento al precio original del producto.

*Al precio ya descontado, sumarle un 19% correspondiente al IVA (multiplicándolo por 1.19).

*Devolver (return) el precio final de venta.

*Llama a tu función simulando un producto que cuesta $30000 y tiene un descuento del 20%, e imprime el resultado.

```python
def aplicar_descuento_iva (precio_producto, porcentaje_descuento):
  descuento = precio_producto * porcentaje_descuento / 100
  print("el descuento es de", descuento)
  precio_con_descuento = precio_producto - descuento
  print("el precio on descuento es de", precio_con_descuento)
  precio_con_iva = precio_con_descuento * 1.19
  return precio_con_iva
```

```python
print("el precio total con iva es de", aplicar_descuento_iva(30000, 20))
```

2.Realizar un validador de contraseñas

Solicite una contraseña al usuario, Valide según multiples condiciones, Maneje errores/excepciones.

*Pedir la contraseña por input.

*Validar que tenga almenos 8 caracteres. Si no se cumple, imprimir "Error: Minimo 8 caracteres".

*Contener al menos un número. Si no se cumple, imprimir "Error: Falta numero".

*Incluir al menos una mayúscula. Si no se cumple, imprimir "Error: Falta mayuscula".

*Si todas las condiciones se cumplen, imprimir "Clave valida".

Implementar estructuras if-elif-else, Manejo de excepciones con try-except, Uso de raise para errores personalizados.

```python
try:

  contraseña = input("Ingrese su clave: ").strip()

  if len(contraseña) < 8:
    raise ValueError ("Minimo 8 caracteres.")
  if not any(c.isdigit() for c in contraseña):
    raise ValueError ("Falta numero")
  if not any(c.isupper() for c in contraseña):
    raise ValueError ("Falta mayuscula")
  else:
    print("Clave valida")

except ValueError as e:
  print(f"Error: {e}")
```
