```python
## Variables ##

nombre_producto = input("ingrese el nombre del producto")
precio_producto = input("ingrese el precio unitario del producto")
cantidad_producto = input("ingrese la cantidad vendidad del producto")

total_venta = float(precio_producto) * int(cantidad_producto)

print("usted compro", cantidad_producto,"", nombre_producto, "a un precio unitario de", precio_producto, "lo que da un total de", total_venta)
```

```python
## Estructuras de Datos ##

categorias_tienda = ("Frutas", "Verduras", "Lacteos", "Carnes", "Bebidas")
compra_casa_1 = {"nombre_producto": "manzana", "categoria": "Frutas", "precio_unitario": 1000, "cantidad": 4, "total": 4000}
compra_casa_2 = {"nombre_producto": "Kumis", "categoria": "Lacteos", "precio_unitario": 6000, "cantidad": 2, "total": 12000}
compra_casa_3 = {"nombre_producto": "carne de res", "categoria": "Carnes", "precio_unitario": 19000, "cantidad": 6, "total": 114000}

registro_ventas = [compra_casa_1, compra_casa_2, compra_casa_3]
```

```python
## Estructuras de Control (Condicionales y Ciclos) ##

for venta in registro_ventas:
    if venta["total"] > 100000:
        print("El total de", venta["nombre_producto"], "Aplica para un descuento del 10%")
        descuento = venta["total"] * 0.10
        total_con_descuento = venta["total"] - descuento
        print("El total con descuento es:", total_con_descuento)
    else:
        print("El total de", venta["nombre_producto"], "No Aplica para un descuento")
```

```python
## Funciones y Manejo de Errores ##

def calcular_ingreso(precio, cantidad):
    try:
        ingreso = int(precio) * int(cantidad)
        return ingreso
    except (TypeError, ValueError):
        print("Error: Los valores deben ser numéricos")
    else:
        print("El ingreso calculado es:", ingreso)
    finally:
        print("Cálculo de ingreso finalizado")

resultado1 = calcular_ingreso(100, 5)
print("Ingreso:", resultado1)

resultado2 = calcular_ingreso("abc", 5)
print("Ingreso:", resultado2)
```
